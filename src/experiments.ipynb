```python
import torch

def fx_initial_params_to_tensor(config, device="cpu", dtype=torch.float32):
    """
    Convert grouped FX config JSON into a flat tensor of 49 parameters
    in the exact order expected by FXChain.process_normalized().

    Args:
        config (list): JSON array with 3 dicts: EQ, Compressor, Reverb
        device (str): torch device
        dtype: torch dtype

    Returns:
        torch.Tensor of shape (49,)
    """

    # -----------------------------
    # Extract blocks by type
    # -----------------------------
    eq = next(c for c in config if c["type"] == "EQ")
    comp = next(c for c in config if c["type"] == "Compressor")
    rev = next(c for c in config if c["type"] == "Reverb")

    params = []

    # -----------------------------
    # 1) EQ (18)
    # -----------------------------
    for i in range(1, 7):
        params.extend([
            eq[f"b{i}_freq"],
            eq[f"b{i}_gain"],
            eq[f"b{i}_q"],
        ])

    # -----------------------------
    # 2) Compressor (6)
    # -----------------------------
    params.extend([
        comp["threshold_db"],
        comp["ratio"],
        comp["attack"],
        comp["release"],
        comp["makeup_gain_db"],
        comp["mix"],
    ])

    # -----------------------------
    # 3) Reverb (25)
    # -----------------------------
    params.extend([
        # Early
        rev["early_gain"],
        rev["early_delay"],
        rev["early_diffusion"],
        rev["early_width"],
        rev["early_lowcut"],
        rev["early_highcut"],
        rev["early_mix"],

        # Late
        rev["late_gain"],
        rev["decay_time"],
        rev["late_diffusion"],
        rev["density"],
        rev["mod_rate"],
        rev["mod_depth"],
        rev["late_lowcut"],
        rev["late_highcut"],
        rev["late_width"],
        rev["late_mix"],

        # Global / Output
        rev["pre_delay"],
        rev["damping"],
        rev["lowcut"],
        rev["highcut"],
        rev["wet"],
        rev["dry"],
        rev["width"],
        rev["mix"],
    ])

    assert len(params) == 49, f"Expected 49 params, got {len(params)}"

    return torch.tensor(params, device=device, dtype=dtype).unsqueeze(0)
```
