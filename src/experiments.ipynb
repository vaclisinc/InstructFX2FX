## Set Instructions

```python
llmclient = LLMClient()
fx_chain_factory = FXChainFactory()
fx_chain = fx_chain_factory.create_fx_chain(sample_rate=44100, device=device)

text_anchor = "This sound is dark"
text_target = "This sound is bright"
task = "Make the sound brighter"
instruction = "This is piano music. " + task

prompt = PromptFactory.LLM_PARAMETER_INITIALIZATION_PROMPT_DASP(fx_chain, instruction)
```

```python
print(f"\n🤖 Method 1: LLM generates initial parameters")
print(f"Instruction to LLM: '{method1_config.prompt.instruction}'")
# Generate initial parameters that will be used for ALL experiments
if not llm_params_tensor_example or not llm_params_dict_example:
    llm_params = llmclient.generate_parameters(method1_config.prompt)
    llm_params_initial = fx_initial_params_to_tensor(llm_params, device=device, ALL_PARAM_RANGES=ALL_PARAM_RANGES)
    print(f"✓ LLM generated {llm_params_initial.shape[1]} parameters")
    print(f"Sample: {llm_params_initial[0, :5].tolist()}...")
else:
    llm_params_initial = fx_initial_params_to_tensor(llm_params_dict_example, device=device, ALL_PARAM_RANGES=ALL_PARAM_RANGES)
    llm_params_dict_initial = llm_params_dict_example
```

```python
try:
    # Apply LLM params
    audio_llm_initial = fx_chain(audio, llm_params_initial)
    print("\n🎵 Audio with initial LLM parameters:")
    display(play_audio(audio_llm_initial))
except Exception as e:
    print(f"⚠️ Could not apply LLM parameters to audio: {e}")
    print("This may be due to incompatible parameter ranges or other issues. Proceeding to refinement step without listening to the initial LLM output.")
```

## Method 2: LLM init, Text2FX loss + GD

```python
method2_config = Config(
    prompt=prompt,
    device=device,
    initialization_method=ParameterInitializationMethod.LLM,
    optimization_method=OptimizationMethod.GRADIENT_DESCENT,
    num_iterations=100,
    learning_rate=0.01,
    fx_chain=fx_chain
)
```

```python
print("\n🎯 Method 2: Text2FX Refinement (LLM init)")
print(f"Prompt: '{prompt.instruction}'")

trainer2 = Trainer(config=method2_config)

params_refined_llm, history_llm, snapshots_llm = trainer2.train(
    audio=audio,
    embedding=clap,
    initial_params=llm_params_initial,
    text_anchor=text_anchor,
    text_target=text_target,
)

audio_refined_llm = fx_chain(audio, torch.sigmoid(params_refined_llm))
print("\n🎵 Final refined audio (LLM init + Text2FX):")
play_audio(audio_refined_llm)
```

## Method 3: Random init, Text2FX & GD

```python
method3_config = Config(
    prompt=prompt,
    device=device,
    initialization_method=ParameterInitializationMethod.RANDOM,
    optimization_method=OptimizationMethod.GRADIENT_DESCENT,
    num_iterations=100,
    learning_rate=0.01,
    fx_chain=fx_chain
)
```

```python
print("\n🎲 Step 3: Text2FX with Random init")
print("Starting from: Random parameters")

trainer3 = Trainer(config=method3_config)
random_params = torch.randn_like(llm_params_initial)

params_refined_random, history_random, snapshots_random = trainer3.train(
    audio=audio,
    embedding=clap,
    initial_params=random_params,
    text_anchor=text_anchor,
    text_target=text_target,
)

audio_refined_random = fx_chain(audio, torch.sigmoid(params_refined_random))
print("\n🎵 Final refined audio (Random init + Text2FX):")
play_audio(audio_refined_random)
```
